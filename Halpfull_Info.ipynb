# Variable Definitions

## `country`
**Description:** Country where the interviewee was surveyed.

**Type:** Categorical

**Possible Values:**
- Kenya
- Rwanda
- Tanzania
- Uganda

**Why it matters:**
Different countries may have different levels of financial inclusion, banking infrastructure, economic development, and access to financial services.

---

## `year`
**Description:** Year in which the survey was conducted.

**Type:** Numerical (Discrete)

**Example Values:**
- 2016
- 2017
- 2018

**Why it matters:**
Financial inclusion may have changed over time due to economic growth, government programs, or increased adoption of mobile banking.

---

## `uniqueid`
**Description:** Unique identifier assigned to each interviewee.

**Type:** Identifier

**Example Values:**
- `uniqueid_1`
- `uniqueid_1456`

**Why it matters:**
This variable is only used to uniquely identify respondents and should be removed before training a machine learning model because it contains no predictive information.

---

## `location_type`
**Description:** Type of area where the interviewee lives.

**Type:** Categorical

**Possible Values:**
- `Urban`
- `Rural`

**Why it matters:**
People living in urban areas often have better access to banks and financial services than people living in rural areas.

---

## `cellphone_access`
**Description:** Indicates whether the interviewee has access to a mobile phone.

**Type:** Categorical (Binary)

**Possible Values:**
- `Yes`
- `No`

**Why it matters:**
Access to a mobile phone is strongly related to financial inclusion because mobile banking and mobile money services are common in many African countries.

---

## `household_size`
**Description:** Number of people living in the same household.

**Type:** Numerical

**Example Values:**
- 1
- 4
- 8

**Why it matters:**
Household size may influence income distribution, expenses, saving behavior, and access to financial services.

---

## `age_of_respondent`
**Description:** Age of the interviewee.

**Type:** Numerical

**Example Values:**
- 18
- 35
- 62

**Why it matters:**
Financial behavior often varies by age. Younger and older individuals may have different probabilities of owning a bank account.

---

## `gender_of_respondent`
**Description:** Gender of the interviewee.

**Type:** Categorical

**Possible Values:**
- `Male`
- `Female`

**Why it matters:**
Gender differences may influence access to banking services and financial opportunities.

---

## `relationship_with_head`
**Description:** Relationship of the interviewee to the head of the household.

**Type:** Categorical

**Possible Values:**
- `Head of Household`
- `Spouse`
- `Child`
- `Parent`
- `Other relative`
- `Other non-relatives`
- `Dont know`

**Why it matters:**
A person's role within the household may affect their financial independence and likelihood of owning a bank account.

---

## `marital_status`
**Description:** Marital status of the interviewee.

**Type:** Categorical

**Possible Values:**
- `Married/Living together`
- `Single/Never Married`
- `Divorced/Separated`
- `Widowed`
- `Dont know`

**Why it matters:**
Marital status may influence financial responsibilities, income stability, and banking behavior.

---

## `education_level`
**Description:** Highest level of education completed by the interviewee.

**Type:** Categorical

**Possible Values:**
- `No formal education`
- `Primary education`
- `Secondary education`
- `Vocational/Specialised training`
- `Tertiary education`
- `Other/Dont know/RTA`

**Why it matters:**
Education is often one of the strongest predictors of financial inclusion because it is linked to employment opportunities, income, and financial literacy.

---

## `job_type`
**Description:** Main occupation or income source of the interviewee.

**Type:** Categorical

**Possible Values:**
- `Farming and Fishing`
- `Self employed`
- `Formally employed Government`
- `Formally employed Private`
- `Informally employed`
- `Remittance Dependent`
- `Government Dependent`
- `Other Income`
- `No Income`
- `Dont Know/Refuse to answer`

**Why it matters:**
Employment status and income source strongly affect a person's need for and ability to access banking services.

---

## Target Variable: `bank_account`

**Description:** Indicates whether the interviewee has or uses a bank account.

**Type:** Categorical (Binary)

**Possible Values:**
- `Yes`
- `No`

**Why it matters:**
This is the variable the machine learning model will predict. The goal of the project is to identify individuals who are most likely to have a bank account and understand the factors influencing financial inclusion.
